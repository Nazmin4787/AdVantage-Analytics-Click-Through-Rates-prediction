## 6. Statistical tests

The tests and plots in Sections 6 and 7 use a copy of the data with missing values filled by the median. This full-data imputation is only used for these checks. The model itself imputes inside a pipeline after the train/test split, so no test-set information leaks into training.


```python
# Copy used only for the statistical tests, correlation and outlier plots.
# (The model itself imputes inside the Pipeline, after the train/test split.)
data_simple_imputed = data.copy()
for col in ["Age", "Time_Spent_on_Site", "Previous_Clicks"]:
    data_simple_imputed[col] = data_simple_imputed[col].fillna(data_simple_imputed[col].median())

data_simple_imputed.isna().sum()
```

### 6.1 Device type vs. clicked (chi-square test)

**Question:** is clicking independent of the device type?


```python
# Contingency table: Device_Type x Clicked
contingency_table = pd.crosstab(data_simple_imputed['Device_Type'], data_simple_imputed['Clicked'])

chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Test of Independence for Device_Type and Clicked:")
print("---------------------------------------------------------")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print("\nContingency Table:")
print(contingency_table)

if p_value < 0.05:
    print("\nConclusion: Statistically significant association found. The type of device a user is on is strongly related to whether they click on an ad.")
else:
    print("\nConclusion: No statistically significant association. Device type does not appear to be related to ad clicks.")
```

**Inference**

- The chi-square statistic is 5,830.6 with p < 0.001, so device type and clicking are strongly associated.
- Mobile: 5,901 clicks and **0** non-clicks. Desktop: 903 of 3,096 clicked (29.2%). Tablet: 314 of 1,003 clicked (31.3%).
- **Business reading:** mobile impressions are the most valuable inventory in this data, so prioritising mobile delivery would raise CTR.
- **Caution:** a device where every single impression is clicked is a red flag for synthetic data or a logging artifact. Confirm with the data owner before acting on it.


### 6.2 Previous clicks vs. clicked (Mann-Whitney U test)

**Question:** do users who clicked have a different history of past clicks than users who did not? The Mann-Whitney test is used because `Previous_Clicks` is a skewed count.


```python
# Split Previous_Clicks by outcome
previous_clicks_clicked = data_simple_imputed[data_simple_imputed['Clicked'] == 1]['Previous_Clicks'].dropna()
previous_clicks_not_clicked = data_simple_imputed[data_simple_imputed['Clicked'] == 0]['Previous_Clicks'].dropna()

u_stat, p_value = mannwhitneyu(previous_clicks_clicked, previous_clicks_not_clicked, alternative='two-sided')

print("Mann-Whitney U Test for Previous_Clicks:")
print("------------------------------------------")
print(f"U-statistic: {u_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("\nConclusion: Statistically significant difference found. The distribution of previous clicks is significantly different for users who clicked versus those who did not.")
else:
    print("\nConclusion: No statistically significant difference. The number of previous clicks does not significantly differ between users who clicked and those who did not.")
```

**Inference**

- p = 0.38, which is above 0.05, so there is no evidence that past clicks differ between clickers and non-clickers.
- In this dataset, click history does not predict the current click. This goes against the usual assumption that engaged users keep clicking, so it is worth checking how the field was recorded.
- The feature is kept in the model, but its effect is expected to be close to zero.

## 8. Modelling

### 8.1 Features and stratified split

- **Predictors (10):** Age, Time_of_Day, Time_Spent_on_Site, Previous_Clicks, Gender, Geographic_Location, Ad_Position, Device_Type, Ad_Category, Day_of_Week.
- **Excluded:** `User_ID` and `Ad_ID` (identifiers).
- **Split:** 80% train / 20% test, stratified on `Clicked` so both sets keep the same click rate.
- The split happens **before** any imputation or encoding is learned, so the test set stays unseen.


```python
# Define the predictors available at prediction time
numeric_features = [
    'Age',
    'Time_of_Day',
    'Time_Spent_on_Site',
    'Previous_Clicks'
]

categorical_features = [
    'Gender',
    'Geographic_Location',
    'Ad_Position',
    'Device_Type',
    'Ad_Category',
    'Day_of_Week'
]

features = numeric_features + categorical_features
target = 'Clicked'

# Use the raw data: imputation and encoding happen inside the Pipeline, after the split
X = data[features]
y = data[target]

# Split BEFORE learning imputation values or encoding categories
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution (%):")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting target distribution (%):")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

### 8.2 Preprocessing and model pipeline

| Step | Columns | What it does |
|---|---|---|
| Median imputation | Numeric (4) | Fills missing values with the training median |
| Most-frequent imputation | Categorical (6) | Fills missing values with the training mode |
| One-hot encoding | Categorical (6) | One 0/1 column per category; unseen categories are ignored |
| Logistic Regression | All features | `liblinear` solver, L2 regularisation (default `C=1`), `class_weight="balanced"` |

With `class_weight="balanced"` on the training data, each non-click counts about 1.73 times and each click about 0.70 times. The numeric features are **not scaled**, so their coefficients are per raw unit (per year, per minute, per click).

Everything is wrapped in one `Pipeline`, so the same steps are applied to training and test data and nothing is learned from the test set.


```python
# Numeric columns: learn medians from training data only
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median'))
])

# Categorical columns: learn frequent values and categories
# from training data only
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Apply the appropriate preprocessing to each column group
preprocessor = ColumnTransformer(transformers=[
    ('numeric', numeric_transformer, numeric_features),
    ('categorical', categorical_transformer, categorical_features)
])

# Combine preprocessing and Logistic Regression into one pipeline
log_reg_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        solver='liblinear',
        random_state=42,
        class_weight='balanced',    #class_weight='balanced' makes sure that the model is not biased towards one class and dishes out a higher penalty for misclassifying the minority class.
        max_iter=1000
    ))
])

print("Preprocessing and Logistic Regression pipeline created.")
```

### 8.3 Training and coefficients


```python
# Train the whole pipeline (imputation + encoding + Logistic Regression) on the training data only
log_reg_model.fit(X_train, y_train)

feature_names = log_reg_model.named_steps['preprocessor'].get_feature_names_out()

print("Original input features:", X_train.shape[1])
print("Features after preprocessing:", len(feature_names))
print("Model training complete.")
print("Intercept:", log_reg_model.named_steps['classifier'].intercept_)
```

```python
# Coefficient (log-odds change) for every encoded feature
coefficients = log_reg_model.named_steps['classifier'].coef_[0]

coef_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

print(coef_df)
```

**Inference**

- **Device dominates the model.** Mobile has a coefficient of +5.53, Desktop -2.80 and Tablet -2.67. Every level of every other categorical feature lies within about ±0.16.
- The gap between Mobile and Desktop is 8.33 log-odds, an odds ratio of roughly 4,100. That is an unrealistically large effect. It comes from near-perfect separation (Mobile is never a non-click). L2 regularisation keeps the coefficient finite, but the size should not be quoted as a real effect size.
- Smaller effects point the expected way: Top Banner (+0.15) beats Side Panel (-0.15), and Education (+0.15) beats E-commerce (-0.16). Whether they matter in practice is untested.
- Numeric coefficients are per raw unit, so they cannot be compared directly with the dummy variables. Time on site adds +0.025 per minute (odds x1.025 per minute, about x1.29 per 10 minutes). Previous clicks add +0.022 per click, Age +0.001 per year, and hour of day -0.008 per hour.
- Because every category level is kept (no reference level dropped), compare levels **within** the same feature.


**Odds ratios.** The exponent of a coefficient is the factor by which the odds of a click change. A value above 1 raises the odds, and a value below 1 lowers them. The table is sorted by the size of the coefficient.


```python
# Odds ratio = exp(coefficient); sorted by absolute coefficient size
coef_df['Odds_Ratio'] = np.exp(coef_df['Coefficient'])
odds_table = coef_df.reindex(coef_df['Coefficient'].abs().sort_values(ascending=False).index)

print(odds_table.round(3).to_string(index=False))
```

### 9.3 Threshold tuning (0.30)

A click is predicted when the probability is at least the threshold. Lowering it finds more clickers but adds false alarms; raising it does the opposite.


```python
# Get probability of class 1 (Clicked)
y_proba = log_reg_model.predict_proba(X_test)[:, 1]

# Set your threshold here
threshold = 0.30  # a click is predicted when the probability is at or above the threshold

# Convert probabilities into predictions
y_pred_threshold = (y_proba >= threshold).astype(int)

print(confusion_matrix(y_test, y_pred_threshold))
print(classification_report(y_test, y_pred_threshold))
print(accuracy_score(y_test,y_pred_threshold))
```

**Inference**

| Threshold | True clicks found | False alarms | Missed clickers | Accuracy |
|---|---|---|---|---|
| 0.50 | 1,172 | 0 | 252 | 0.874 |
| 0.30 | 1,177 | 3 | 247 | 0.875 |

- Lowering the threshold from 0.50 to 0.30 changed only 8 predictions (5 correct clicks gained, 3 false alarms), because almost no user scores between those values.
- Reaching the remaining 247 clickers would likely mean flagging nearly all non-Mobile users, where only about 30% click. That is no better than the overall base rate for that group.
- **Rule of thumb:** lower the threshold when a missed clicker costs more than a wasted impression, and raise it when impressions are expensive. In a real project the threshold should be chosen on validation data and the business cost of each error, not on the test set.

## 10. Conclusions, limitations and next steps

### Key findings

1. **Device type is the only strong driver of clicks.** Mobile impressions always clicked; Desktop and Tablet click about 30% of the time.
2. **Other features have small effects.** Ad position (Side Panel is lowest) and ad category (Gaming and Education are highest) show small CTR gaps; gender, location and day of week are not meaningful drivers.
3. **Click history does not predict clicks here** (Mann-Whitney p = 0.38).
4. **The model beats the baseline** (accuracy 0.874 vs 0.712) with perfect precision on clicks, but almost all of that comes from the Mobile pattern.

### Recommendations

- Prioritise Mobile inventory, but first confirm the 100% Mobile click rate with the data owner. If it is real, it is a major finding; if it is an artifact, the model's results do not carry over to real traffic.
- Do not rely on this model for Desktop and Tablet users. Add richer signals such as ad creative, bid price, referrer, session depth and past CTR per ad.
- Test Top Banner / Footer placements against Side Panel in an A/B experiment before changing inventory allocation.

### Limitations

- One train/test split only, with no cross-validation or confidence intervals.
- ROC-AUC and precision-recall curves are not reported.
- Numeric features are unscaled and the regularisation strength (`C=1`) was not tuned.
- The threshold was examined on the test set. This is acceptable for a demonstration but not for production.
- Coefficients show association, not cause and effect.
- The data shows signs of being synthetic (a device with a 100% click rate and uniform distributions).

### Next steps

1. Fit a device-only model and compare it with the full model to measure what the other features add.
2. Add ROC-AUC, a precision-recall curve and 5-fold cross-validation.
3. Scale numeric features and tune `C`.
4. Try interaction terms (for example device x ad position) and a tree-based model for comparison.
5. Check probability calibration before using scores for bidding.


## Appendix A: Feature engineering experiment (not used in the final model)

Four extra features were created to see whether they add signal: `User_Engagement_Score`, `Peak_Time_Ad_Flag` (a **weekend** flag), `Ad_Relevance_Score` and `Repeat_User_Flag`. A VIF check was then run.

The final model in Section 8 uses only the original features, for these reasons:

- **Infinite VIF.** `User_Engagement_Score` is exactly `Time_Spent_on_Site + Previous_Clicks`, a perfect linear combination of two other features, so all three show an infinite VIF. The remaining features have VIF between 1.0 and 1.7.
- **No meaningful unit.** Adding minutes to a click count does not describe anything real.
- **Possible leakage.** `Ad_Relevance_Score` was assigned from CTRs computed on the full dataset, so it would let information about the target into the features.
- `User_ID` and `Ad_ID` appear in the VIF table only because they are numeric columns; they are not model features.


```python
# Load the imputed dataset
data_fe = data_simple_imputed.copy()

# --- 1. User Engagement Score ---
# A score combining Time_Spent_on_Site and Previous_Clicks
# We'll use a simple approach: (Time_Spent_on_Site + Previous_Clicks)
data_fe['User_Engagement_Score'] = data_fe['Time_Spent_on_Site'] + data_fe['Previous_Clicks']

# --- 2. Peak Time Ad Flag ---
# A binary flag for weekend days, which may have different user behavior
data_fe['Day_of_Week_encoded'] = data_fe['Day_of_Week'].map({
    'Mon': 1, 'Tue': 2, 'Wed': 3, 'Thu': 4, 'Fri': 5, 'Sat': 6, 'Sun': 7
})
data_fe['Peak_Time_Ad_Flag'] = np.where(
    data_fe['Day_of_Week_encoded'].isin([6, 7]),  # 6=Saturday, 7=Sunday
    1, 0
)

# --- 3. Ad Relevance Score ---
# A simple categorical score based on our EDA insights
ad_relevance_mapping = {
    'Gaming': 'High', 'Education': 'High',
    'Travel': 'Medium', 'Finance': 'Medium',
    'E-commerce': 'Low'
}
data_fe['Ad_Relevance_Score'] = data_fe['Ad_Category'].map(ad_relevance_mapping)

# --- 4. Repeat User Flag ---
# A binary variable that flags users with a history of clicks
data_fe['Repeat_User_Flag'] = np.where(
    data_fe['Previous_Clicks'] > 0,
    1, 0
)

# Preview the new features clearly
print("Preview of Business-Centric Engineered Features:")
print(data_fe[
    ['User_Engagement_Score', 'Peak_Time_Ad_Flag',
     'Ad_Relevance_Score', 'Repeat_User_Flag']
].head())

# To continue your project, you should save this new dataframe
# data_fe.to_csv('ad_click_data_engineered.csv', index=False)
```
